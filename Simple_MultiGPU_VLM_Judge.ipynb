# VLM-as-a-Judge: Evaluate Multimodal Models

Use a Vision-Language Model to judge image captioning and VQA quality.

- **Captioning Model:** BLIP (base, generates captions)
- **Judge Model:** Qwen2-VL-2B-Instruct (sees image + caption, scores quality)
- **Method:** Multi-GPU parallel inference with DeepSpeed
- **Criteria:** Accuracy, Detail, Relevance, Fluency
- **Platform:** Kaggle 2x T4 (free) or Colab
