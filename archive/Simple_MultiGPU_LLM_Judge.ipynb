# LLM-as-a-Judge: Evaluate Fine-Tuned Models

Use a judge LLM to score model responses on multiple criteria.

- **Target Model:** Qwen2.5-0.5B + LoRA (fine-tuned via DPO)
- **Judge Model:** Qwen2.5-1.5B-Instruct (bigger = better judge)
- **Method:** Multi-GPU parallel judging with DeepSpeed ZeRO-2
- **Criteria:** Helpfulness, Coherence, Safety, Creativity
- **Platform:** Kaggle 2x T4 (free) or Colab
